```python
import BAC0
import pandas as pd
import time
from datetime import datetime

# Initialize BACnet
bacnet = BAC0.connect("192.168.1.95/24")

# Define BACnet device IP address
device_ip = '192.168.1.60'

# Define the points to be read
points = [
    'analogValue 0',  # Replace with your actual object identifiers
    'analogValue 1',
    'analogValue 2'
]

header_ = [
    'Tag1',  # Replace with your actual object identifiers
    'Tag2',
    'Tag3'
]

# Function to generate CSV file name based on current date
def generate_csv_file_name():
    current_date = datetime.now().strftime('%Y-%m-%d')
    return f'bacnet_data_log_{current_date}.csv'

# Get initial CSV file name
csv_file = generate_csv_file_name()

# Create an empty DataFrame with the desired columns
df = pd.DataFrame(columns=['Timestamp'] + header_)

# Variable to keep track of the current date
current_date = datetime.now().date()

try:
    while True:
        # Get current timestamp and date
        timestamp = datetime.now().strftime('%Y-%m-%d %H:%M:%S')
        new_date = datetime.now().date()
        
        # Check if the date has changed
        if new_date != current_date:
            # Update the current date
            current_date = new_date
            
            # Generate new CSV file name
            csv_file = generate_csv_file_name()
            
            # Create a new DataFrame for the new day
            df = pd.DataFrame(columns=['Timestamp'] + header_)
        
        # Read data from each point
        data_row = [timestamp]
        for point in points:
            try:
                # Construct the BACnet read command
                value = bacnet.read(f'{device_ip} {point} presentValue')
                data_row.append(value)
            except Exception as e:
                data_row.append(None)  # Append None if there's an error
        
        # Append the data row to the DataFrame
        df.loc[len(df)] = data_row
        
        # Print the logged data (optional)
        # print(f'Logged data: {data_row}')
        
        # Write the DataFrame to the CSV file
        df.to_csv(csv_file, index=False)
        
        # Wait for 2 seconds before next read
        time.sleep(2)
except KeyboardInterrupt:
    print("Logging stopped by user.")
```
