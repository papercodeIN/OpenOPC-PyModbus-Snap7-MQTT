```python
#!/usr/bin/env python
# coding: utf-8

import random
from time import sleep
from BAC0 import connect
from BAC0.core.devices.local import factory
from BAC0.core.devices.local.factory import (
    ObjectFactory,
    analog_input,
    analog_output,
    binary_output,
    binary_input,
)

# ------------------------------------------------------------------------------
# Function to define BACnet objects
# ------------------------------------------------------------------------------
def defining_objects(device):
    # Clear existing objects before creating new ones
    ObjectFactory.clear_objects()

    analog_input(
        instance=1,
        name="Frequency",
        properties={"units": "hertz"},
        description="Frequency sensor",
        presentValue=18.0,
    )
    analog_input(
        instance=2,
        name="Barometer",
        properties={"units": "pascals"},
        description="Room Pressure sensor",
        presentValue=101325.0,
    )
    analog_input(
        instance=3,
        name="Humidity",
        properties={"units": "percentRelativeHumidity"},
        description="Room Humidity sensor",
        presentValue=45.0,
        relinquish_default=45.0,
    )
    analog_input(
        instance=4,
        name="Temperature",
        properties={"units": "degreesCelsius"},
        description="Room Temperature sensor",
        presentValue=22.0,
        relinquish_default=22.0,
    )
    analog_output(
        instance=5,
        name="GasResistance",
        properties={"units": "ohms"},
        description="Gas Resistance output",
        presentValue=200.0,
        relinquish_default=200.0,
    )
    binary_output(
        instance=6,
        name="RoomOneHeatingEnabled",
        description="Room one heating enabled",
        presentValue=True,
    )
    binary_output(
        instance=7,
        name="RoomTwoHeatingEnabled",
        description="Room two heating enabled",
        presentValue=True,
    )
    binary_input(
        instance=8,
        name="RoomOneRadiatorState",
        description="Room one radiator on/off",
        presentValue=False,
    )
    binary_input(
        instance=9,
        name="RoomTwoRadiatorState",
        description="Room two radiator on/off",
        presentValue=False,
    )

    # Register objects in this BACnet device
    factory.add_objects_to_application(device.app)
```

```python
# ------------------------------------------------------------------------------
# Main
# ------------------------------------------------------------------------------
if __name__ == "__main__":
    try:
        # Create BACnet device
        device1 = connect(ip="192.168.1.10/24", port=47808, deviceId=1110)
        defining_objects(device1)
        # logging.info("+ BACnet simulated device started (deviceId=1110).")

        # Simulation loop
        while True:
            print("in while")
            try:
                # Update values dynamically
                device1["Frequency"].presentValue = random.uniform(45, 55)  # Hz
                device1["Temperature"].presentValue = random.uniform(20, 26)  # °C
                device1["Humidity"].presentValue = random.uniform(40, 60)  # %

                # logging.info(
                #     f"Frequency={device1['Frequency'].presentValue:.2f} Hz | "
                #     f"Temperature={device1['Temperature'].presentValue:.2f} °C | "
                #     f"Humidity={device1['Humidity'].presentValue:.2f} %"
                # )

                sleep(2)
            except Exception as loop_err:
                print(f"Simulation loop error: {loop_err}")
                sleep(2)

    except KeyboardInterrupt:
        print("Simulation stopped by user (Ctrl+C).")
    except Exception as e:
        print(f"Failed to start BACnet simulator: {e}")
```

```python
#!/usr/bin/env python
# coding: utf-8

import random
from time import sleep
from BAC0 import connect
from BAC0.core.devices.local.factory import analog_input, analog_output, binary_output

def defining_objects(device):
    analog_input(name="Frequency", instance=1, description="Frequency", properties={"units": "hertz"}, presentValue=18.0)
    analog_input(name="Temperature", instance=2, description="Room Temperature", properties={"units": "degreesCelsius"}, presentValue=22.0)
    binary_output(name="HeatingEnabled", instance=3, description="Heating On/Off", properties={"activeText": "On", "inactiveText": "Off"}, presentValue=True)
    # Register all objects
    analog_input.add_objects_to_application(device)

if __name__ == "__main__":
    device = connect(ip="192.168.1.10/24", port=47808, deviceId=1110)
    
    defining_objects(device)
    print("BACnet simulated device started.")

    while True:
        device["Frequency"].presentValue = random.uniform(45, 55)
        device["Temperature"].presentValue = random.uniform(20, 26)
        device["HeatingEnabled"].presentValue = random.choice([True, False])
        
        print(f"Freq: {device['Frequency'].presentValue:.2f} Hz | Temp: {device['Temperature'].presentValue:.2f} °C | Heating: {device['HeatingEnabled'].presentValue}")
        sleep(2)
```
