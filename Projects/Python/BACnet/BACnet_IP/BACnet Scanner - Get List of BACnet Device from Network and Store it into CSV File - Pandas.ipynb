# Import necessary libraries

```python
import pandas as pd                 # Data analysis and manipulation library
import BAC0                         # Main library for discovering and interacting with BACnet devices
```

# Define the IP range to scan for BACnet devices. Adjust this according to your network.

```python
ip_range = "192.168.1.12/24"  
csv_filename = "bacnet_devices_pandas.csv"  # Name of the CSV file to store the discovered device information
```

# Function to discover BACnet devices on the specified network range

```python
def discover_bacnet_devices(ip_range):
    # Initialize BAC0 with the provided IP range. This sets up a BACnet network scanner.
    bacnet = BAC0.lite(ip_range)  
    
    # Discover BACnet devices on the network
    bacnet.discover()  
    
    # Retrieve the discovered devices from the BACnet instance
    devices = bacnet.devices  
    return devices  # Return the list of discovered devices
```

# Function to save the discovered devices' information into a CSV file

```python
def save_to_csv(devices, csv_filename):
    # Prepare a list to store device information in a structured format
    device_data = []
    for device in devices:
        # Extract relevant information from each discovered device
        device_info = {
            "Device Name": device[0],        # Assuming index 0 contains the device name
            "Device Instance": device[3],    # Assuming index 3 contains the device instance ID
            "MAC Address": device[2],        # Assuming index 2 contains the MAC address
            "Model Name": device[1]          # Assuming index 1 contains the model name
        }
        # Add the device information to the list
        device_data.append(device_info)
    
    # Convert the list of device dictionaries into a DataFrame
    df = pd.DataFrame(device_data)  
    
    # Save the DataFrame to a CSV file
    df.to_csv(csv_filename, index=False)  
```

# Main block: runs the discovery and CSV saving functions when the script is executed

```python
if __name__ == "__main__":
    # Discover BACnet devices
    devices = discover_bacnet_devices(ip_range)  
    
    # Save the discovered devices to a CSV file
    save_to_csv(devices, csv_filename)  
    
    # Print a confirmation message
    print(f"BACnet devices discovered and saved to '{csv_filename}'.")
```
