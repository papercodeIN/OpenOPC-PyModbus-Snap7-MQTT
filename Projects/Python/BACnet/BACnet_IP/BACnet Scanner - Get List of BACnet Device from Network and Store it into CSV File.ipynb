# Import Libraries for Data Handling and BACnet Interaction

```python
import csv  # Library to handle CSV file operations
import BAC0  # Library to interact with BACnet devices
```

# Setup the IP Address and CSV File Name to Store the Device Data

```python
ip_range = "192.168.1.10/24"  # Define the IP range for BACnet devices discovery (you can change this to your network range)
csv_filename = "bacnet_devices.csv"  # Define the output CSV file where device data will be stored
```

# Discover the BACnet Devices in the IP Range

```python
def discover_bacnet_devices(ip_range):
    """
    Discover BACnet devices on the given IP range using BAC0.
    
    Args:
        ip_range (str): The IP range to scan for BACnet devices.
        
    Returns:
        devices (list): A list of BACnet devices found in the network.
    """
    bacnet = BAC0.connect(ip_range)  # Connect to the BACnet network using the IP range
    bacnet.discover()  # Discover BACnet devices on the network
    devices = bacnet.devices  # Retrieve the discovered devices from the BACnet network
    return devices  # Return the list of discovered devices
```

# Save Device Details to CSV File

```python
def save_to_csv(devices, csv_filename):
    """
    Save the details of discovered BACnet devices to a CSV file.
    
    Args:
        devices (list): List of BACnet devices to save to the file.
        csv_filename (str): The name of the CSV file where the data will be saved.
    """
    # Open the CSV file in write mode ('w') and prepare to write data
    with open(csv_filename, mode='w', newline='') as file:
        writer = csv.writer(file)  # Create a CSV writer object
        # Write the header row with column names
        writer.writerow(["Device Name", "Device Instance", "Device IP", "Model Name"])

        # Iterate over the discovered devices and write their details to the CSV file
        for device in devices:
            device_name = device[0]  # Extract device name
            device_instance = device[3]  # Extract device instance (unique identifier)
            device_ip = device[2]  # Extract device IP address
            model_name = device[1]  # Extract model name (assuming it's stored in index 1)
            # Write the device data in the corresponding CSV row
            writer.writerow([device_name, device_instance, device_ip, model_name])
```

# Run All The Functions Created Above to Perform the Action

```python
if __name__ == "__main__":
    # Discover BACnet devices in the specified IP range
    devices = discover_bacnet_devices(ip_range)
    
    # Save the discovered devices to the specified CSV file
    save_to_csv(devices, csv_filename)
    
    # Print confirmation message indicating the operation was successful
    print(f"BACnet devices discovered and saved to '{csv_filename}'.")
```
